```python
import os
import sys
import pandas as pd
from pathlib import Path

# --- 0. RESET WORKING DIRECTORY & CLEANUP ---
# Move back to base content directory first so bash doesn't get stranded
%cd /content
!rm -rf /content/ILs-screening-tm
!rm -rf /content/ils_screening_tm

# --- 1. CLONE THE GITHUB REPOSITORY ---
GITHUB_URL = "https://github.com/Toucouere-Lorris/ILs-screening-tm.git"
REPO_DIR = Path("/content/ILs-screening-tm")

print(f"Attempting to clone: {GITHUB_URL}")
!git clone {GITHUB_URL} {REPO_DIR}

# --- 2. VERIFICATION ---
if not REPO_DIR.exists():
    print("\n❌ Cloning failed. Please verify that the repository visibility is set to public.")
else:
    print("\n✓ Repository successfully cloned! Configuring environment...")

    # --- 3. INSTALL DEPENDENCIES & PACKAGE ---
    %cd {REPO_DIR}
    print("\nInstalling the package and its dependencies (RDKit, TensorFlow, Pandas...)...")
    !pip install -e .
    print("✓ Dependencies installed.")

    # --- 4. CONFIGURE PYTHON SEARCH PATHS ---
    if str(REPO_DIR) not in sys.path:
        sys.path.insert(0, str(REPO_DIR))

    package_dir = REPO_DIR / "ils_screening_tm"
    if str(package_dir) not in sys.path:
        sys.path.insert(0, str(package_dir))

    # --- 5. ALIGN DATA DIRECTORY STRUCTURE ---
    COLAB_BASE = Path("/content")
    os.makedirs(COLAB_BASE / "output", exist_ok=True)

    if (package_dir / "database").exists() and not (COLAB_BASE / "database").exists():
        !cp -r {package_dir / "database"} {COLAB_BASE}
        print("✓ 'database/' directory copied to Colab root for environment compatibility.")

    # --- 6. LAUNCH INTERACTIVE DASHBOARD ---
    print("\nLoading interactive screening dashboard...")
    try:
        from ils_screening_tm.main import start_screening_interface

        base_cations_path = COLAB_BASE / "database" / "base_cations.csv"
        if base_cations_path.exists():
            df_scaffolds = pd.read_csv(base_cations_path)
            print(f"✓ {len(df_scaffolds)} scaffolds loaded. Displaying dashboard:")
            print("-" * 60)

            # Execute ipywidgets interface
            start_screening_interface(df_scaffolds)
        else:
            print(f"❌ Error: base_cations.csv not found at {base_cations_path}")
    except Exception as e:
        print(f"❌ Error during interface execution: {e}")
```
